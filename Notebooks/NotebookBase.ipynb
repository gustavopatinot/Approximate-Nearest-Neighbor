```python
import platform
import sys
import subprocess
import os
import psutil
import json
from datetime import datetime, timezone

def safe_command(command: str) -> str:
    try:
        return subprocess.check_output(command, shell=True, text=True).strip()
    except Exception as exc:
        return f"ERROR: {exc}"

def capture_runtime_environment():
    env = {
        "timestamp_utc": datetime.now(timezone.utc).isoformat(),
        "python_version": sys.version,
        "python_executable": sys.executable,
        "platform": platform.platform(),
        "machine": platform.machine(),
        "processor": platform.processor(),
        "cpu_count_logical": psutil.cpu_count(logical=True),
        "cpu_count_physical": psutil.cpu_count(logical=False),
        "ram_total_gb": round(psutil.virtual_memory().total / (1024**3), 2),
        "current_working_directory": os.getcwd(),
        "git_commit": safe_command("git rev-parse HEAD"),
        "git_branch": safe_command("git branch --show-current"),
        "pip_freeze": safe_command("pip freeze"),
        "nvidia_smi": safe_command("nvidia-smi"),
    }
    return env

runtime_environment = capture_runtime_environment()

print(json.dumps(
    {k: v for k, v in runtime_environment.items() if k != "pip_freeze"},
    indent=2
))
```

```python
def get_git_info(command):
    import subprocess
    try:
        return subprocess.check_output(command, shell=True, text=True).strip()
    except Exception:
        return "unknown"

current_git_commit = get_git_info("git rev-parse HEAD")
current_git_branch = get_git_info("git branch --show-current")

run_id = f"{experiment_config['experiment']['id']}_{datetime.now(timezone.utc).strftime('%Y%m%d_%H%M%S')}_{uuid.uuid4().hex[:8]}"
manifest = {
    "run_id": run_id,
    "git_commit": current_git_commit,
    "git_branch": current_git_branch,
    "base_config": base_config,
    "experiment_config": experiment_config,
    "hardware": hardware,
}

save_json(manifest, f"results/logs/{run_id}_manifest.json")

print(f"Run manifest saved: {run_id}")
```

```python
import pandas as pd
from datetime import datetime, timezone

query_results = []

query_results.append({
    "run_id": run_id,
    "experiment_id": experiment_config["experiment"]["id"],
    "method": "fcvi",
    "dataset": "synthetic_1m",
    "seed": 42,
    "query_id": 0,
    "k": 10,
    "filter_type": "tenant_id",
    "filter_selectivity": 0.01,
    "filter_cardinality": 100,
    "hard_filter_present": True,
    "soft_filter_present": False,
    "alpha": 1.5,
    "lambda": 0.7,
    "k_prime": 100,
    "ef_search": 100,
    "recall_at_10": 0.92,
    "precision_at_10": 0.90,
    "ndcg_at_10": 0.88,
    "filtered_recall_at_10": 0.91,
    "violation_rate": 0.0,
    "constraint_satisfaction_rate": 1.0,
    "latency_ms": 42.7,
    "queue_wait_ms": 0.0,
    "service_time_ms": 42.7,
    "search_time_ms": 30.1,
    "rerank_time_ms": 8.2,
    "validation_time_ms": 1.5,
    "memory_mb": 2048.0,
    "timestamp_utc": datetime.now(timezone.utc).isoformat()
})

df = pd.DataFrame(query_results)

output_path = f"results/raw/{run_id}_query_results.csv"
df.to_csv(output_path, index=False)

print("Saved:", output_path)
```
